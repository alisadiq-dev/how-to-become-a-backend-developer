> # Topic: Magic methods & @property

> ### 1) Quick explanation: Magic Methods & Properties 
>
> * **Magic Methods:** Special methods that start and end with `__` (double underscore). Python calls them **automatically**.
>   * *Purpose:* They let your objects behave like built-in types (working with `print()`, `==`, `len()`, etc.).
>
> * **Key Examples:**
>   * `__str__`: Returns **human-friendly** text (used by `print()`).
>   * `__repr__`: Returns **developer-friendly** text (used for debugging/logs).
>   * `__eq__`: Defines how `==` compares two objects.
>   * `__len__`: Defines what `len(obj)` returns.
>
> * **Properties (`@property`):**
>   * Lets you access a method like a variable (e.g., `obj.name` instead of `obj.get_name()`).
>   * **`@setter`**: Controls how data is saved to that attribute (great for validation/safety).

> ### 2) Real-world analogy / visual
> **Think of a box**
>
> ```text
> print(box)     -> label on the box        (__str__)
> box == other   -> same contents?          (__eq__)
> len(box)       -> how many items inside   (__len__)
> box.price      -> price tag you see       (@property)
> box.price = x  -> cashier checks value    (@setter)
> ```

````python
class Task:
    def __init__(self, title: str, done: bool = False):
        self.title = title
        self.done = done

    def __str__(self):
        return f"Task: {self.title} (done={self.done})"

    def __repr__(self):
        return f"Task(title={self.title!r}, done={self.done})"

    def __eq__(self, other):
        if not isinstance(other, Task):
            return False
        return self.title == other.title and self.done == other.done

    def __len__(self):
        # Length of the title
        return len(self.title)


t1 = Task("Write code")
t2 = Task("Write code")

print(t1)            
print(repr(t1))       
print(t1 == t2)       
print(len(t1))      
````

> ### example (@property and @setter)

````python
class BankAccount:
    def __init__(self, balance: float):
        self._balance = balance  # underscore = internal use

    @property
    def balance(self):
        # Read access
        return self._balance

    @balance.setter
    def balance(self, value):
        # Write access with validation
        if value < 0:
            raise ValueError("Balance cannot be negative")
        self._balance = value


acc = BankAccount(100.0)
print(acc.balance)      # looks like an attribute

acc.balance = 150.0     # validated write
print(acc.balance)

# acc.balance = -10.0   # would raise ValueError
````
