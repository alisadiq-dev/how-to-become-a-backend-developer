#  Debugging & Common OOP Pitfalls

Even experienced developers make mistakes in Object-Oriented Programming. Here are **6 common mistakes** beginners make in Python, why they happen, and **how to fix them**.

---

### 1️ The "Mutable Default Argument" Trap
**The Mistake:** Using a mutable object (like a list or dictionary) as a default argument.
**Why it fails:** Python creates the default object **once** (when the function is defined), not every time you call it. Everyone ends up sharing the same list!



** Wrong:**
```python
class Student:
    def __init__(self, name, grades=[]):   
        self.name = name
        self.grades = grades

s1 = Student("Ali")
s2 = Student("Sara")
s1.grades.append(90)

print(s2.grades)  

````python
# correct 
class Student:
    def __init__(self, name, grades=None):
        self.name = name
        self.grades = grades if grades is not None else []  
````

### 2️ Class vs. Instance Attributes Confusion

**The Mistake:** Modifying a class attribute thinking it will only change for one specific object.
**Why it fails:** Class attributes are shared by **all** instances of that class. If you change a mutable class attribute (like a list), that change is reflected in every single object!



####  Wrong (Shared Data)
```python
class WebBrowser:
    history = []   
    def visit(self, url):
        self.history.append(url)

user1 = WebBrowser()
user2 = WebBrowser()

user1.visit("google.com")

#   User2 sees User1's history
print(f"User2 History: {user2.history}")


### correct 

````python
class WebBrowser:
    def __init__(self):
        self.history = []  # <--- Instance attribute (Unique to THIS object)

    def visit(self, url):
        self.history.append(url)

user1 = WebBrowser()
user2 = WebBrowser()

user1.visit("google.com")

 
print(f"User2 History: {user2.history}")
````

### 3️ Forgetting `self` in Methods

**The Mistake:** Defining a method inside a class without `self` as the first parameter.
**Why it fails:** When you call `obj.method()`, Python **automatically** passes the object (`obj`) as the first argument. If your function isn't defined to accept it, Python raises a `TypeError`.



####  Wrong (Missing Argument)
```python
class Dog:
    def bark():  # <--- Missing 'self' here!
        print("Woof!")


# Python tries to call Dog.bark(instance), but bark() takes 0 arguments.
# Dog().bark()  
# TypeError: bark() takes 0 positional arguments but 1 was given

````python
# correct 
class Dog:
    def bark(self):  # <--- 'self' catches the instance
        print("Woof!")

d = Dog()
d.bark()  # Works perfectly!
````

### 4️ Not Calling `super().__init__`

**The Mistake:** Overriding the `__init__` method in a child class but forgetting to initialize the parent class.
**Why it fails:** The parent class's setup code never runs, so inherited attributes (like `name` in this example) are never created.



####  Wrong (Parent Skipped)
```python
class Employee:
    def __init__(self, name):
        self.name = name

class Manager(Employee):
    def __init__(self, name, dept):
        self.dept = dept
        #  Problem: Employee.__init__ was never called!
        # self.name is missing.

# m = Manager("Ali", "IT")
# print(m.name)  
# AttributeError: 'Manager' object has no attribute 'name'

````python
# correct 
class Manager(Employee):
    def __init__(self, name, dept):
        
        super().__init__(name)  
        self.dept = dept

m = Manager("Ali", "IT")
print(m.name)  
````

### 5️ Bloated `__init__` Methods

**The Mistake:** Putting too much logic (calculations, database connections, printing) inside `__init__`.
**Why it fails:** It makes the object hard to test and slow to create. `__init__` should just assign values.



####  Wrong (Doing too much)
```python
class DataAnalyzer:
    def __init__(self, data):
        self.data = data
        print("Analyzing...")       
         

    def heavy_calculation(self):
        # ... complex math ...
        pass

````python
# correct 
class DataAnalyzer:
    def __init__(self, data):
        self.data = data  

    def analyze(self):     
        return self.heavy_calculation()
````

### 6️ Misusing `is` vs `==`

**The Mistake:** Using `is` to compare values (like strings, numbers, or lists) instead of `==`.

**Why it fails:**
* **`is`** checks if two objects are the **exact same object in memory** (Identity).
* **`==`** checks if their **values** are equal (Equality).



####  Wrong (Checking Identity)
```python
a = [1, 2, 3]
b = [1, 2, 3]

# They look the same, but they live in different memory locations
print(f"Is same object? {a is b}")
# Output: False

````python
# correct 
# To check if contents are the same, use ==
print(f"Has same value? {a == b}")
# Output: True
````
