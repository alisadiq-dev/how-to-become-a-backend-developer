> ## Explain lists, tuples, sets, and dictionaries

> ### 1) List
>
> **What it is:**
> A list is an **ordered**, **changeable** collection of items.
>
> **When to use it:**
> * When order matters
> * When you need to add, remove, or change items
> * When items can repeat

````python
# Example 
numbers = [1, 2, 3, 4]
print(numbers[0])   
````

````python
# Example 
fruits = ["apple", "banana"]
fruits.append("mango")
print(fruits)     
````

> ### 2) Tuple
>
> **What it is:**
> A tuple is an **ordered**, **unchangeable** collection of items.
>
> **When to use it:**
> * When data should **not** change
> * For fixed values (coordinates, settings)

````python
# Example 
point = (10, 20)
print(point[0])     
````

````python
colors = ("red", "green", "blue")
# colors[0] = "black"  not allowed
print(colors)
````

> ### 3) Set
>
> **What it is:**
> A set is an **unordered** collection of **unique** items.
>
> **When to use it:**
> * When you want **no duplicates**
> * For membership checks (**fast lookups**)

````python
numbers = {1, 2, 2, 3}
print(numbers)      
````

````python
users = {"ali", "ahmed"}
print("ali" in users)  
````

> ### 4) Dictionary
>
> **What it is:**
> A dictionary stores data as **key : value** pairs.
>
> **When to use it:**
> * When data has names/labels
> * When you want **fast access** by key

````python
student = {"name": "Ali", "age": 20}
print(student["name"])   
````

````python
marks = {}
marks["math"] = 90
marks["english"] = 85
print(marks)             
````

````python
# problem : If you want to store student name → marks, which one fits best?
marks = {
    "Ali": 85,
    "Ahmed": 92
}

print(marks["Ali"])   
````

> ### how to iterate dictionaries and use dictionary methods (get, keys, values).

````python
# 1) Sample dictionary

marks = {
    "Ali": 85,
    "Ahmed": 92,
    "Sara": 88
}
for name in marks:
    print(name)
````

````python
# b) Loop over values
for score in marks.values():
    print(score)
````

````python
# c) Loop over key + value together (MOST IMPORTANT)
for name, score in marks.items():
    print(name, "got", score)
````

> #### Remember this:

> ** items() → key + value

> ** This is used a lot in backend work

> ## 3) Dictionary methods

````python
print(marks.get("Ali"))      
print(marks.get("Usman"))    
````

````python
print(marks.get("Usman", 0)) 

# Why get() is good:
# No error if key doesn’t exist
# Very useful in real projects
````

> ## b) keys()

````python
print(marks.keys())
````

> ## c) values()

````python
print(marks.values())
````

> ###  Exercise 1: Count occurrences
> **Task:** Create a dictionary that counts how many times each name appears.
>
 

````python
names = ["Ali", "Sara", "Ali", "Ahmed", "Sara", "Ali"]
counts = {}

for name in names:
    # get(name, 0) checks if name exists; if not, returns 0
    counts[name] = counts.get(name, 0) + 1

print(counts)
````

###  Exercise 2: Filter dictionary values
> **Task:** Print only students who scored 90 or above.
>
> ```python


````python
marks = {
    "Ali": 85,
    "Ahmed": 92,
    "Sara": 88,
    "Usman": 75
}

for name, score in marks.items():
    if score >= 90:
        print(name)
````

> ### Exercise 3: Build a new list from a dictionary
> **Task:** Create a list of items that cost more than 100.
>
> ```python


````python
prices = {
    "apple": 120,
    "banana": 60,
    "mango": 150,
    "orange": 80
}

expensive_items = []

for item, price in prices.items():
    if price > 100:
        expensive_items.append(item)

print(expensive_items)


 
````
