## Topic: Control Flow — if / elif / else

> ### 1) Quick explanation
> Programs often need to choose what to do based on conditions.
>
> * **`if`** checks a condition (True / False).
> * **`elif`** means "else if" → check another condition if the previous was False.
> * **`else`** runs when none of the above conditions are True.
>
> *Conditions usually use comparisons like `>`, `<`, `==`, `!=`.*

 > ### 2) Real-world analogy / visual
> **Think of exam results:**
>
> * **If** marks >= 80 → **Grade A**
> * **Else if** marks >= 60 → **Grade B**
> * **Else** → **Fail**
>
> *Only one path runs.*

> ## 3) Example 

````python
marks = 72 
if marks >= 80:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Fail")


````

````python
# example 2
age = 17

if age >= 18:
    print("You can vote")
else:
    print("You cannot vote")
````

## Topic: Loops — for-loop & while-loop

> ### 1) Quick explanation
> Loops let you repeat code instead of writing it again and again.
>
> * **`for` loop** → use when you know how many times to repeat.
> * **`while` loop** → use when you don’t know how many times, only the condition.
>
> *Loops stop when the sequence ends (for) or condition becomes False (while).*

> ### 2) Real-world analogy / visual
>
> **For-loop → Attendance list** 
> * "Check each student in this list one by one"
>
> **While-loop → ATM PIN** 
> * "Keep asking for PIN while it’s wrong"
>
> **Visual:**
> ```text
> FOR:    List → item1 → item2 → item3 → stop
> WHILE:  Condition True → repeat → repeat → False → stop
> ```

> ### Example 1: `for` loop (counting)
> > ### Use `for` when:
> * You know the range.
> * You loop over lists, strings, ranges.

````python
for i in range(1,6):
    print(i)
````

````python
# Example with a list:

names = ["Ali", "Ayesha ", "Ehtisham", "Dua"]
for name in names:
    print(name)
````

> ### Example 2: `while` loop (condition-based)
> > ### Use `while` when:
> * You don’t know exact repetitions
> * Loop depends on a condition changing

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

> ### 4) When to use which? (Very important)
>
> | Situation | Use |
> | :--- | :--- |
> | Looping over list / range | **for** |
> | Fixed number of times | **for** |
> | Condition-based repetition | **while** |
> | User input until valid | **while** |
>
> ---
>
> **Rule of thumb:**
> * If you can say *"for each item"* → use **for**
> * If you can say *"keep going while..."* → use **while**

> #  Exercise: Loop Practice
>
> **Task:**
> 1. Use a **for-loop** to print numbers from **10 to 1**.
> 2. Use a **while-loop** to print numbers from **1 to 5**.

> ###  Explanation: Backwards Loop
>
> ```python
> for number in range(10, 0, -1):
>     print(number)
> ```
>
> **How `range(start, stop, step)` works here:**
>
> * **`start = 10`** → Begin from **10**.
> * **`stop = 0`** → End *before* **0** (so it stops at 1).
> * **`step = -1`** → Go **backwards** (decrease by 1).

> ###  Explanation: While Loop
>
> ```python
> number = 1
>
> while number <= 5:
>     print(number)
>     number += 1
> ```
>
> **Breakdown:**
> * **`number = 1`** → Starting point.
> * **`while number <= 5`** → Loop runs while condition is `True`.
> * **`number += 1`** → Increases number by 1 each time (**VERY important**).

# Topic: List Comprehensions

> ### 1) Quick explanation
> A list comprehension is a short way to build a list.
> * It replaces a `for` loop that appends items to a list.

### Normal for-loop:

````python
squares = []
for x in range(1, 6):
    squares.append(x * x)
````

### List comprehension (same result, one line):

````python
squares = [x * x for x in range(1, 6)]
````

> #### Example 

````python
numbers = [x for x in range(5)]
print(numbers)
````

````python
names = ["ali", "sara", "ahmed"]
upper_names = [name.upper() for name in names]
print(upper_names)
````

> #### Example 3: Using if (filtering)

````python
nums = [1, 2, 3, 4, 5, 6]
even_nums = [n for n in nums if n % 2 == 0]
print(even_nums)
````

> #  Exercise: List Comprehension
>
> **Task:**
> 1. Use a list comprehension to create **squares** of numbers **1–10**.
> 2. From `[5, 12, 7, 18, 3]`, create a new list with numbers **greater than 10**.
> 3. From `["apple", "banana", "cherry"]`, create a list of **word lengths**

````python
 # 1. Squares of 1-10
squares = [x**2 for x in range(1, 11)]
print(f"Squares: {squares}")
````

````python
# 2. Filter numbers > 10
nums = [5, 12, 7, 18, 3]
big_nums = [n for n in nums if n > 10]
print(f"Numbers > 10: {big_nums}")
````

````python
# 3. Word lengths
fruits = ["apple", "banana", "cherry"]
lengths = [len(word) for word in fruits]
print(f"Lengths: {lengths}")
````
