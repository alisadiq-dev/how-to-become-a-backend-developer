> # Python Variables & Data Types

> ### 1) Variables (what they are)
> A variable is just a name that stores a value.
>
> * **the label** = variable name
> * **the thing inside** = value
>
> ```python
> age = 20
> ```
>
> **Here:**
> * `age` → variable name
> * `20` → value

> ### 2) Data Types in Python
> Python automatically understands the type of data you store.
>
> **We’ll cover 4 basic types:**
>
> * **`int`** → whole numbers
> * **`float`** → decimal numbers
> * **`str`** → text
> * **`bool`** → True / False

> ### 3) Very Simple Examples 
>
> #### Example 1: `int` (Integer – whole number)
> ```python
> x = 10
> print(x)
> ```
> **Output:**
> ```text
> 10
> ```
>
> **What’s happening?**
> * `x` stores a whole number → **int**

> #### Example 2: `float` (Decimal number)
> ```python
> price = 99.5
> print(price)
> ```
> **Output:**
> ```text
> 99.5
> ```
>
> **What’s happening?**
> * `price` stores a decimal number → **float**

> #### Example 3: `str` 
> ```python
> name = "Ali"
> print(name)
> ```
> **Output:**
> ```text
> Ali
> ```
>
> **What’s happening?**
> * Text is written inside quotes `" "` → **str**

````python
is_student = True 
print(is_student)
````

### Data type Example 

````python
x = 10
y = 3.5
name = "Ali"
is_student = True

print(type(x))
print(type(y))
print(type(name))
print(type(is_student))
````

> ## Question
> Show how to convert between common types (int↔str, str↔float) with runnable examples

> ### 1) Quick explanation
> Sometimes data is in the wrong type.
> Python gives us built-in functions to convert:
>
> * **`int()`** → make an integer
> * **`float()`** → make a decimal number
> * **`str()`** → make text
>
> *Conversion works only if the value makes sense.*

> ### 2) Real-world analogy
> **Think of it like changing containers** 
>
> ```text
> "25"   →   int("25")   →   25
> (text)     (convert)      (number)
> ```

> ### 3) Runnable examples
>
> #### A) int ↔ str

````python
age = 20
age_text = str(age)

print(age)
print(age_text)
print(type(age))
print(type(age_text))
````

````python
# Example 2 
marks_text = "85"
marks = int(marks_text)

print(marks)
print(type(marks))
````

## B) str ↔ float

````python
price_text = "99.5"
price = float(price_text)

print(price)
print(type(price))
````

````python
pi = 3.14
pi_text = str(pi)

print(pi_text)
print(type(pi_text))
````

> ### 4) Very important common mistakes 
>
> ** Mistake 1: Invalid string to int**
> ```python
> x = int("10.5")
> #  Error because "10.5" is not an int
> ```
> ** Fix:
> ```python
> x = float("10.5")
> ```
>
> ** Mistake 2: Letters inside string**
> ```python
> y = int("20a")
> #  Error: Python cannot convert letters
> ```
> **Fix:**
> * Remove letters
> * Or keep it as `str`
>
> ** Mistake 3: Thinking conversion changes original value**
> ```python
> a = "30"
> int(a)
> print(a)
> ```
> **Output:**
> ```text
> 30
> ```
> 👉 `a` is still a string
>
> ** Correct way:**
> ```python
> a = int(a)
> ```

## basic string operations (concatenation, f-strings, .format, slicing)

> ### 1) Quick explanation
> * **Concatenation** → join strings using `+`
> * **f-strings** → modern, clean way to insert values
> * **.format()** → older but still common formatting style
> * **Slicing** → take a part of a string using indexes

> ### 2) Runnable examples
>
> #### A) String concatenation (+)
> **Concept:** Joining two or more strings together using the `+` sign.
>
> ```python
> first = "Data"
> last = "Science"
>
> # Combine them (don't forget the space!)
> full_name = first + " " + last
> print(full_name)
> ```
> **Output:**
> ```text
> Data Science
> ```

````python
first = "Ali"
last = "Sadiq"

full_name = first + " " + last
print(full_name)
````

## f-strings

````python
name = "Ali"
age = 20

msg = f"My name is {name} and I am {age} years old"
print(msg)
````

## .format() method

````python
name = "Ali"
age = 20

msg = "My name is {} and I am {} years old".format(name, age)
print(msg)
````

## String slicing

````python
word = "Python"

print(word[0])      # first character
print(word[0:3])    # from index 0 to 2
print(word[-1])     # last character
````

> ### Visual Aid: String Indexing
> Python counts positions starting from **0**, not 1.
>
> ```text
>  P   y   t   h   o   n
>  0   1   2   3   4   5
> ```
