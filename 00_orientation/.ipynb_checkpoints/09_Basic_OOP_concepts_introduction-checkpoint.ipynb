> # Topic: Basic OOP — Classes & Objects

> ### 1) Quick explanation: Classes & Objects 
>
> * **Class:** A **blueprint** (plan). It describes what something is and what it can do.
> * **Object:** A **real thing** made from the blueprint.
> * **Structure:** The class defines **data** (variables) and **behavior** (functions, called **methods**).
> * **Key Concept:** Each object has its **own** data, even though all objects use the same class.

> ### 2) Real-world analogy / visual
> **Think of a bank form vs actual accounts:**
>
> ```text
> Class: BankAccount (blueprint)
>  ├── owner
>  ├── balance
>  ├── deposit()
>  └── withdraw()
>
> Objects (real accounts):
>  ├── Ali's account → balance = 1000
>  └── Sara's account → balance = 500
> ```
> * **Conclusion:** Same blueprint, different real accounts.

````python
# Example 
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount
        print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Not enough balance!")
        else:
            self.balance = self.balance - amount
            print(f"Withdrew {amount}. New balance: {self.balance}")

account1 = BankAccount("Ali", 1000)
account2 = BankAccount("Sara", 500)

account1.deposit(200)
account1.withdraw(300)

account2.withdraw(700)
account2.deposit(100)
````

````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"{self.owner} deposited ${amount}. New Balance: ${self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds!")
        else:
            self.balance -= amount
            print(f"{self.owner} withdrew ${amount}. New Balance: ${self.balance}")
````

> # Topic: __init__, instance vs class attributes

Goal: Understand how __init__ works, the difference between instance attributes and class attributes, and how multiple objects are created from one class.

> ### 1) Quick explanation: The `__init__` Method 🛠️
>
> * **`__init__`:** A special method that runs **automatically** when you create an object.
>   * *Purpose:* It is used to set up **initial data** for each object.
>
> * **Instance Attributes:** Belong to **one specific object** (unique data).
> * **Class Attributes:** Belong to the **class itself** and are **shared** by all objects.
>
> * **Key Concept:** One class → you can create **many objects**, each with its own data.

> ### 2) Real-world analogy / visual
> **Think of a bank rule vs individual accounts:**
>
> ```text
> Class attribute (shared rule):
> - bank_name = "ABC Bank"
>
> Instance attributes (per account):
> - owner
> - balance
>
> Account 1 (Ali)      Account 2 (Sara)
> owner = "Ali"        owner = "Sara"
> balance = 1000       balance = 500
> bank_name = ABC Bank bank_name = ABC Bank
> ```
> * **Conclusion:** Same bank, different people and balances.

````python
# Example 
class BankAccount:
    bank_name = "Meezan Bank"
    def __init__(self, owner, balance):
        
        self.owner = owner
        self.balance = balance

    def show_details(self):
        print(f"Bank: {BankAccount.bank_name}")
        print(f"Owner: {self.owner}")
        print(f"Balance: {self.balance}")
        print(" ")



account1 = BankAccount("Ali", 1000)
account2 = BankAccount("Ehtisham", 500)
account3 = BankAccount("Dua", 2000)

account1.show_details()
account2.show_details()
account3.show_details()
````

> #  Exercise: Class Attributes vs Instance Attributes
>
> **Task:**
> 1. Create a `BankAccount` class with a shared `bank_name` and individual `owner`/`balance`.
> 2. Create two initial accounts.
> 3. **Change the `bank_name`** for the whole class.
> 4. Create a **new account**.
> 5. Print details to see what happened.
> * **Did bank_name change for all accounts?**
>   * **YES.** Even though `Ali` and `Sara` were created *before* the change, their bank name updated automatically because it is a **shared link**.
> * **Did owners/balances stay different?**
>   * **YES.** Changing the bank name did not touch their personal money or names.
>   * 

````python
class BankAccount:
    bank_name = "HBL Bank"

    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

acc1 = BankAccount("Ali", 1000)
acc2 = BankAccount("Sara", 500)

print(f"Before Update: {acc1.owner} uses {acc1.bank_name}")
BankAccount.bank_name = "National Bank of pakistan"
acc3 = BankAccount("adeen", 2000)
print(f"{acc1.owner}: {acc1.bank_name} (Balance: {acc1.balance})")
print(f"{acc2.owner}: {acc2.bank_name} (Balance: {acc2.balance})")
print(f"{acc3.owner}: {acc3.bank_name} (Balance: {acc3.balance})")
````

> # Topic: Methods and self

> ### 1) Quick explanation: Methods & `self` 
>
> * **Method:** A function that lives **inside a class**.
> * **`self`:** Means **"this object"** (the specific object calling the method).
> * **Mechanism:** Python passes the object **automatically** into the method as the first argument (`self`).
> * **Purpose:** `self` lets the method **read/change** that specific object’s own data.
>
> ** Memory Trick:**
> * `self` = **current object**

````python
# example 
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def check_balance(self):
        print(f"{self.owner}'s balance is {self.balance}")

    def deposit(self, amount):
        self.balance = self.balance + amount
        print(f"{self.owner} deposited {amount}")

account1 = BankAccount("Ali", 1000)
account2 = BankAccount("Sara", 500)

account1.check_balance()
account2.check_balance()

account1.deposit(200)
account1.check_balance()
````
