> # Topic: Functions (Basics)

> ### 1) Quick explanation
>
> * **Function:** A named block of code you can reuse.
> * **Parameters:** Variables written in the function definition.
> * **Arguments:** The real values you pass when calling the function.
> * **Return value:** What the function sends back using `return`.
> * **Default arguments:** Give parameters a fallback value if you don’t pass one.

> ### 2) Real-world analogy / visual
> **Think of a function like a juice machine **
>
> ```text
> Ingredients (arguments) ──▶ [ FUNCTION ] ──▶ Juice (return value)
> ```
>
> * **Machine recipe** = function definition
> * **Ingredients you put in** = arguments
> * **Juice you get** = return value

> ### 3) Example 

````python
# 1. Define a function
def add(a, b):
    return a + b

# 2. Call the function
result = add(2, 3)
print(result)
````

````python
def greet(name):      # name = parameter
    print("Hello", name)

greet("Ali")          # "Ali" = argument
````

````python
def square(x):
    return x * x

value = square(4)
print(value)
````

````python
def greet(name="Guest"):
    print("Hello", name)

greet("Ali")
greet()
````

> # Topic: Function Scope (Local vs Global)

> ### 1) Quick explanation
>
> * **Local scope:** variables created **inside** a function → usable *only* there.
> * **Global scope:** variables created **outside** all functions → usable *everywhere*.
> * **Note:** Functions **cannot** change a global variable by default.
>   * *Reason:* Python protects you from accidentally breaking global data.

````python
# Example 
def greet():
    message = "Hello"   # local variable
    print(message)

greet()
# print(message)  ERROR
````

````python
count = 0   # global

def increment():
    count = count + 1   # ❌ mistake
    print(count)

increment()
````

````python
# Fix 1: Return the value (BEST practice)
count = 0

def increment(count):
    return count + 1

count = increment(count)
print(count)
````

````python
#Fix 2: Using global keyword
count = 0

def increment():
    global count
    count = count + 1

increment()
print(count)
````

> #  Exercise: Scope Practice
>
> **Task:**
> 1. Create a global variable `score = 10`.
> 2. Write a function `add_points(score, points)`.
> 3. Return the updated score.
> 4. Print the final score.
>
> ---
>
> ###  Solution
> ```python
> score = 10  # Global variable
>
> # Function receives 'score' as a parameter (Local scope)
> def add_points(score, points):
>     updated_score = score + points
>     return updated_score
>
> # Call the function and save the result
> # We pass the global 'score' (10) and add 5 points
> new_score = add_points(score, 5)
>
> print(f"Original Global Score: {score}")
> print(f"New Score returned: {new_score}")
> ```
>
> **Expected Output:**
> ```text
> Original Global Score: 10
> New Score returned: 15
> ```
>
> **💡 Key Takeaway:**
> * The `score` inside the function is a **local** copy.
> * The calculation does not change the **global** `score` automatically. It returns a *new* value which you must capture in a variable (like `new_score`).

> # Topic: Modules & Imports (Basics)

> ### 1) Quick explanation
>
> * **Module:** A module is just a Python file (`.py`) with code in it.
> * **Usage:** You use imports to reuse code from one file in another.
> * **Common styles:**
>   * `import module`
>   * `from module import thing`
> * **Search Path:** Python looks for modules in the **same folder** (and a few other places).

> ### 2) Real-world analogy / visual
> **Think of modules like toolboxes **
>
> ```text
> math_utils.py  ──▶  tools (functions)
> main.py        ──▶  uses those tools
> ```
>
> * You don’t rewrite tools every time — you **import** them.

> ### 3) Minimal runnable example
>
> **Step 1: Create a module file**
> * **File:** `math_utils.py`
>
> ```python
> def add(a, b):
>     return a + b
>
> def multiply(a, b):
>     return a * b
> ```
>
> ---
>
> **Step 2: Use the module from another file**
> * **File:** `main.py`
>
> ```python
> import math_utils
>
> result1 = math_utils.add(2, 3)
> result2 = math_utils.multiply(4, 5)
>
> print(result1)
> print(result2)
> ```
>
> **Expected Output:**
> ```text
> 5
> 20
> ```
