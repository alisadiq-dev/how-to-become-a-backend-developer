> # Topic: File I/O (Reading & Writing Text Files)

> ### 1) Quick explanation: File I/O 
>
> **Definition:** File I/O means saving data to a file and reading data back.
> * Python uses `open()` to work with files.
>
> **Common modes:**
> * `"w"` → **write** (creates new file or **overwrites** existing one) 
> * `"r"` → **read** (opens existing file)
> * `"a"` → **append** (adds data at the end of the file)
>
> **Best Practice:**
> * Always prefer: `with open(...) as f:`
> * **Reason:** It **auto-closes** the file safely (even if errors occur).
>
> **Note:** Files are just plain text unless you choose another format (like JSON or CSV).

> ### 2) Real-world analogy / visual
> **Think of a file like a notebook**
>
> ```text
> open file  -> open notebook
> write()    -> write lines with a pen
> read()     -> read what's written
> close()    -> close the notebook
> ```

````python
# Step 1: WRITE to a file
with open("notes.txt", "w") as file:
    file.write("Hello Python\n")
    file.write("This is file I/O\n")

# Step 2: READ from the file
with open("notes.txt", "r") as file:
    content = file.read()

print("File content:")
print(content)
````

> # Exercise: File I/O
>
> **Task:**
> 1. Write your **name** and **age** to a file named `info.txt`.
> 2. Read the file back.
> 3. Print the content to the screen.
>
> ---
>
> ### ✅ Solution
> ```python
> # Step 1: Write to the file
> # "w" mode creates the file or overwrites it
> with open("info.txt", "w") as f:
>     f.write("Name: Ali Sadiq\n")
>     f.write("Age: 21\n")
>
> # Step 2: Read from the file
> # "r" mode opens it for reading
> with open("info.txt", "r") as f:
>     content = f.read()
>
> # Step 3: Print result
> print("--- Reading from info.txt ---")
> print(content)
> ```
>
> **Expected Output:**
> ```text
> --- Reading from info.txt ---
> Name: Ali Sadiq
> Age: 21
> ```

````python

with open("info.txt", "w") as f:
    f.write("Name: Ali Sadiq\n")
    f.write("Age: 21\n")

with open("info.txt", "r") as f:
    content = f.read()

# Step 3: Print result
print(" Reading from info.txt")
print(content)
````

> # Topic: File I/O Practice — Line & Word Count

> Give a small exercise: write a script that counts lines/words in a text file and prints results

> #  Exercise: Word & Line Counter
>
> **Task:**
> 1. Create a file named `story.txt` with a few lines of text (3-4 lines).
> 2. Write a script to **read** the file and calculate:
>    * Total number of **lines**.
>    * Total number of **words**.
> 3. Print the results.
>
> **💡 Hints:**
> * Use `.readlines()` to get a list of lines.
> * Use `.split()` on a string to break it into words (e.g., `"Hello World".split()` becomes `['Hello', 'World']`).
>
> ---
>
> ###  Solution
> ```python
> # Step 1: Create the dummy file first
> text_content = """Python is amazing.
> Backend development is challenging but fun.
> I am learning file handling today."""
>
> with open("story.txt", "w") as f:
>     f.write(text_content)
>
> # Step 2: Read and Count
> num_lines = 0
> num_words = 0
>
> with open("story.txt", "r") as f:
>     lines = f.readlines() # Reads all lines into a list
>     num_lines = len(lines)
>
>     for line in lines:
>         words_list = line.split() # Splits line into words
>         num_words += len(words_list)
>
> print(f"Total Lines: {num_lines}")
> print(f"Total Words: {num_words}")
> ```
>
> **Expected Output:**
> ```text
> Total Lines: 3
> Total Words: 14
> ```

````python
text_content = """"" My name is Ali Sadiq, and I'm the students of Punjab University Lahore 
semter 4 ."""

with open("story.txt", "w") as f:
    f.write(text_content)

# Step 2: Read and Count
num_lines = 0
num_words = 0

with open("story.txt", "r") as f:
    lines = f.readlines() 
    num_lines = len(lines)

    for line in lines:
        words_list = line.split()
        num_words += len(words_list)

print(f"Total Lines: {num_lines}")
print(f"Total Words: {num_words}")
````
