 > ## Introduce the Python 3 & OOP Foundations phase:
> * **What I will learn**
> * **Practical goals for this phase (beginner level)**

> ## What this phase is about
> In this phase, you are learning how to think like a backend developer using Python.
>
> * Not frameworks yet.
> * Not databases yet.
>
> First, you learn how to write **clean, structured, reliable Python code**.

> ## What you will learn
> By the end of this phase, you will understand:
>
> ### 1) Modern Python 3 basics
> You will learn how Python is used professionally, not just for small scripts:
> * **Virtual environments** (why every project is isolated)
> * **Clean project structure** (folders, files, modules)
> * **Writing readable functions and files**
> * **Using modern Python features** (f-strings, type hints)

> ### 2) Object-Oriented Programming (OOP)
> This is the **core skill** of this phase.
>
> **You will learn:**
> * What classes and objects really are
> * How to model real-world things (User, Task, Account) as classes
> * How data (attributes) and behavior (methods) live together
> * **Inheritance** (“is-a” relationship)
> * **Composition** (“has-a” relationship)

> ### 3) Writing safer and cleaner code
> You will learn how to make your code robust:
>
> * **Type hints** (to avoid silly bugs)
> * **Dataclasses** (to reduce boilerplate)
> * **Exception handling** (code should not crash silently)
> * **Logging** instead of `print()`

> ### 4) Working with files and data
> Before databases, you must understand:
>
> * **Reading and writing files**
> * **Saving data as JSON**
> * **Loading data back when the program restarts**

> ## Practical goals (what you will actually build)
> This phase is not theory-only. You will build things step by step.
>
> ### Main project of this phase
> You will build a **CLI Task Manager (Command Line App)**:
>
> **It will:**
> * Use classes and OOP properly
> * Save tasks to a JSON file
> * Load tasks when restarted
> * Handle errors safely
> * Use clean project structure
>
> **This single project proves:**
> * You understand Python basics
> * You understand OOP
> * You can build a small but complete system

> ## Skills you should have after this phase
> By the end, you should confidently be able to say:
>
> *  I can create a Python project from scratch
> *  I know when and why to use classes
> *  I can read other people’s Python code without fear
> *  I can organize code into multiple files/modules
> *  I can build small real-world tools (not just exercises)
>
> This is exactly what **FastAPI, databases, and backend frameworks** expect from you next.

> # Question 2: Prerequisites & Setup Guide
>
> ##  Prerequisite Checklist

> ## Prerequisite Checklist (Python + Tools)
>
> ### 1) Python knowledge (minimum)
> You do **NOT** need to be advanced. You’re ready if you know:
>
> * **Variables** (`x = 5`)
> * **if / else**
> * **for loops**
> * **Functions** (`def my_func():`)
> * **Lists & dictionaries** (basic use)

> ### 2️) Python installation (Python 3.11+)
> **Check if Python is installed**
>
> Open Terminal and run:
> ```bash
> python3 --version
> ```
>
> **Expected:**
> ```text
> Python 3.11.x   (or 3.10+ is okay)
> ```

> ### 3️) Virtual Environment (VERY IMPORTANT)
> **Think of a virtual environment as:**
>
> > “A private box for one project’s Python and libraries”
>
> **1. Create a project folder**
> ```bash
> mkdir backend_engineer
> cd backend_engineer
> ```
>
> **2. Create virtual environment**
> ```bash
> python3 -m venv venv
> ```
>
> **3. Activate it**
> * **Mac / Linux:**
>   ```bash
>   source venv/bin/activate
>   ```
> * **Windows:**
>   ```bash
>   venv\Scripts\activate
>   ```
>
> **You should now see:**
> ```text
> (venv) your-name@machine
> ```
>  **This means your environment is active.**

> ### 4️) Code Editor (VS Code)
> **Install Visual Studio Code:**
>  https://code.visualstudio.com/
>
> **Recommended VS Code extensions:**
>
> *  Python (by Microsoft)
> *  Pylance
> *  GitLens (optional but helpful)
>
> **Open your project:**
> ```bash
> code .
> ```

> ### 5️) Git & GitHub (basic setup)
> **Check Git**
> ```bash
> git --version
> ```
> *If not installed:*
>  https://git-scm.com/downloads
>
> **First-time Git config**
> ```bash
> git config --global user.name "Your Name"
> git config --global user.email "your@email.com"
> ```
>
> **Initialize repo**
> ```bash
> git init
> ```
>
> **Create `.gitignore` file and add:**
> ```text
> venv/
> __pycache__/
> .env
> ```
>
> ### 6️) pip (Python package manager)
> **Upgrade pip (recommended):**
> ```bash
> pip install --upgrade pip
> ```
>
> **Check:**
> ```bash
> pip --version
> ```

> ## one-paragraph study plan for this phase,


> ## Phase Roadmap
> In this phase, I will start by refreshing **basic Python syntax** (running files, indentation, comments, variables, and print), then move step-by-step into **core data types** (int, float, string, bool, list, tuple, dict) and how to use them in real code.
>
> Next, I’ll practice **control flow** (if/else, loops) and **functions** to understand how logic is organized. After that, I’ll learn **modules, imports, and virtual environments** so I can structure projects properly.
>
> Once the basics are solid, I’ll transition into **Object-Oriented Programming (OOP)** by learning classes, objects, `__init__`, methods, and attributes, followed by key OOP ideas like **encapsulation, inheritance, and polymorphism** using very small examples.
>
> Finally, I’ll combine everything by building **tiny practice projects** (like a simple account, student, or task system), using **Git** for version control, so by the end of this phase I can confidently read, write, and structure **beginner-level Python backend code**.
