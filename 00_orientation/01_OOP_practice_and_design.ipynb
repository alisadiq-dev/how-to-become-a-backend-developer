> # OOP practice & design

> ### 1) Quick explanation: Inheritance & Encapsulation 
>
> * **Inheritance:** When one class **extends** another. The child class automatically gets (inherits) the parent’s attributes and methods.
> * **Method Overriding:** When a child class **changes** (redefines) how a parent method works for itself.
> * **`super()`:** A function that lets the child class call the **parent’s version** of a method (often used to keep the parent's setup logic).
> * **Encapsulation:** The practice of keeping data **protected** inside a class so it can't be messed up easily from outside.
>   * *Python's Way:* It uses naming conventions like `_name` (internal use) and `__name` (private) along with **properties** to manage access.

> ### 2) Real-world analogy / visual
> **Think of a bank account:**
>
> ```text
> Account (base class)
>  ├─ owner
>  ├─ balance
>  └─ withdraw()
>
> SavingsAccount (child class)
>  ├─ interest_rate
>  └─ withdraw()  ← same name, different rule (Overriding)
> ```
> * **Concept:** Every savings account **is an** account.
> * **Difference:** But a savings account may have **extra rules** (like interest or withdrawal limits).

````python
# example
class Account:
    def __init__(self, owner: str, balance: float):
        self.owner = owner
        self.balance = balance

    def withdraw(self, amount: float) -> None:
        if amount > self.balance:
            print("Not enough balance")
            return
        self.balance -= amount
        print(f"Withdrawn {amount}. Balance = {self.balance}")


class SavingsAccount(Account):
    def __init__(self, owner: str, balance: float, interest_rate: float):
        super().__init__(owner, balance)  # call parent __init__
        self.interest_rate = interest_rate

    def withdraw(self, amount: float) -> None:
        print("SavingsAccount rules applied")
        super().withdraw(amount)  # reuse parent logic
````

````python
acc = Account("Ali", 1000)
sav = SavingsAccount("Sara", 2000, 0.05)

acc.withdraw(200)
sav.withdraw(300)
````

> ### 4) Encapsulation (Private Attributes & Properties)

````python
class Account:
    def __init__(self, owner: str, balance: float):
        self.owner = owner
        self.__balance = balance   # private

    @property
    def balance(self) -> float:
        return self.__balance

    @balance.setter
    def balance(self, value: float) -> None:
        if value < 0:
            raise ValueError("Balance cannot be negative")
        self.__balance = value
````

````python
acc = Account("Ali", 500)

print(acc.balance)     
acc.balance = 800     
acc.balance = -100     
````
