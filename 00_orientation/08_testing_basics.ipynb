> # Testing basics

> ### 1) Quick explanation: Unit Testing
>
> * **Definition:** Unit testing means checking that small pieces of code (units) work **exactly** as expected.
> * **Mechanism:** A test compares the **Actual Output** (what your code did) vs. the **Expected Output** (what it should have done).
> * **Tool:** Python has a built-in library called `unittest` (no need to install anything extra).
> * **Naming:** Tests usually live in files named `test_something.py`.
> * **Goal:** If the test passes  → code is correct (for that specific case).

> ### 2) Real-world analogy / visual
> **Think of a calculator:**
>
> * You press `2 + 3`
> * You expect `5`
> * If calculator shows `7` → something is broken
>
> **Visual Flow:**
> ```text
> Function ──▶ Result
>    │           │
>    └── Test compares ──▶ Expected value
> ```
