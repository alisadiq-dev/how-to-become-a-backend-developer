> # Topic: Exceptions & Debugging (try / except)

> ### 1) Quick explanation: Exception Handling 
>
> * **Exception:** A **runtime error** (something goes wrong while the program runs).
> * **`try`:** Code that **might fail**.
> * **`except`:** What to do **if it fails**.
>
> **Best Practices:**
> * **Catch specific errors** (like `ValueError`), not everything.
> * **Be polite:** Explain the problem clearly, don’t blame the user.

> ### 2) Real-world analogy / visual
> **Think of filling a form**
>
> ```python
> try:
>     Read user's input
> except:
>     "Sorry, that input isn't valid. Please try again."
> ```

> #### 3) Example 

````python
def read_age():
    try:
        age = int(input("Enter your age: "))
        print(f"Your age is {age}")
    except ValueError:
        print("Please enter a valid number for age.")

read_age()
````

> #  Exercise: Safe Division (Exception Handling)
>
> **Task:**
> Create a script `safe_division.py` that:
> 1. Asks the user for **two numbers**.
> 2. Converts them to integers.
> 3. Divides the first by the second.
> 4. **Handles Errors:**
>    * `ValueError`: If user enters text instead of numbers ("Please enter numbers only").
>    * `ZeroDivisionError`: If user divides by zero ("Cannot divide by zero").
>
 

````python
try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    result = num1 / num2
    print(f"Result: {result}")

except ValueError:
    print("Error: Please enter numbers only! ")

except ZeroDivisionError:
    print("Error: You cannot divide by zero! ")
````

> # Topic: Basic Debugging Tips

> ### 1) Quick explanation: Debugging Basics 
>
> * **Print checks:** simple `print()` statements to show variable values while the program runs (the most common debugging way).
> * **Traceback:** Python’s error report that appears when code crashes. It tells you **where** (line number) and **why** (error type) it failed.
> * **assert:** A quick safety check for assumptions. If the condition is False, the program stops immediately.
>   * *Use case:* "I assume this variable is never negative." -> `assert x >= 0`

````python
# A) Print checks
def add_tax(price):
    print("DEBUG: price =", price)
    tax = price * 0.1
    print("DEBUG: tax =", tax)
    return price + tax

print(add_tax(100))
````

````python
# C) Simple assert
def withdraw(balance, amount):
    assert amount >= 0, "Amount must be positive"
    return balance - amount

print(withdraw(100, 20))
````
