# File I/O in Python

File Input/Output (I/O) refers to the reading and writing of files in a file system. Using Python, we can easily work with files by using built-in methods. This notebook covers various aspects of file I/O, including examples and exercises.

## Reading Files

To read a file, you can use the `open()` function along with the `read()`, `readline()`, or `readlines()` methods. Here's how to read a file:

```python
# Example: Reading a file
filename = 'example.txt'

with open(filename, 'r') as file:
    content = file.read()
    print(content
```

## Writing Files

You can write to a file by using the `open()` function in write mode. Here's how to write to a file:

```python
# Example: Writing to a file
filename = 'output.txt'

with open(filename, 'w') as file:
    file.write('Hello, World!')
```

## Exercises

1. Write a program to count the number of words in a file.
2. Write a program to count the number of lines in a file.


```python
# Exercise: Word Counting
filename = 'example.txt'
word_count = 0

with open(filename, 'r') as file:
    for line in file:
        words = line.split()
        word_count += len(words)

print(f'Total words: {word_count}')
```

```python
# Exercise: Line Counting
filename = 'example.txt'
line_count = 0

with open(filename, 'r') as file:
    for line in file:
        line_count += 1

print(f'Total lines: {line_count}')
```
