###  Quiz Answers Key

1.  **B) `[10, 20, 30, 40]`**
    > *Reason:* Lists are mutable objects. Since `y = x` references the same object, appending to `y` also changes `x`.

2.  **C) `__init__`**
    > *Reason:* This is the constructor method used to initialize attributes.

3.  **C) The specific instance (object)**
    > *Reason:* `self` allows the code to access and modify the attributes specific to the object that called the method.

4.  **B) `Reptile`**
    > *Reason:* Class attributes are shared. Changing `Dog.species` updates it for all instances.

5.  **C) Add `__` (double underscore)**
    > *Reason:* This triggers name mangling, making the variable harder to access from outside.

6.  **B) Method Overriding**
    > *Reason:* This allows a child class to provide a specific implementation of a method already defined in the parent.

7.  **A) `15`**
    > *Reason:* `a=3` and `b` uses the default value of `5`. So, `3 * 5 = 15`.

8.  **C) `len()`**
    > *Reason:* `len()` is the built-in function to return the length of an object.

9.  **B) `2 3 4`**
    > *Reason:* The `range()` function stops *before* the last number (exclusive).

10. **B) Class B gets all attributes/methods of Class A**
    > *Reason:* Inheritance means the child class automatically gets access to the parent's behavior.

###  Part 2: Coding Challenge - The Library System

**Task:** Create a class `Library` that manages books.

**Requirements:**
1. **`__init__`**: Initialize an empty list named `books`.
2. **`add_book(title)`**: Add a book title to the list.
3. **`remove_book(title)`**: Remove a book if it exists. If not, print "Book not found".
4. **`show_books()`**: Print all available books.
5. **`search_book(title)`**: Return `True` if book exists, `False` if not.

**Test Logic:**
* Create a library.
* Add "Python 101" and "Harry Potter".
* Remove "Harry Potter".
* Search for "Python 101" (Should be True).
* Show all books.

````python
class Library:
    def __init__(self):
        self.books = []  # Empty list to store books

    def add_book(self, title):
        self.books.append(title)
        print(f" Added: '{title}'")

    def remove_book(self, title):
        if title in self.books:
            self.books.remove(title)
            print(f" Removed: '{title}'")
        else:
            print(f" Book '{title}' not found!")

    def show_books(self):
        print(f"\n Current Library: {self.books}")

    def search_book(self, title):
        if title in self.books:
            return True
        else:
            return False

# --- Testing the Class ---
my_lib = Library()

# 1. Add Books
my_lib.add_book("Python 101")
my_lib.add_book("Harry Potter")
my_lib.add_book("Data Science Handbook")

# 2. Remove a Book
my_lib.remove_book("Harry Potter")

# 3. Show Books
my_lib.show_books()

# 4. Search for a Book
search_query = "Python 101"
found = my_lib.search_book(search_query)

if found:
    print(f"\n '{search_query}' is available!")
else:
    print(f"\n '{search_query}' is NOT available.")
````
